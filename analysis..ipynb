```python
import pandas as pd

df = pd.read_csv("https://football-data.co.uk/mmz4281/2223/E0.csv")

clubs = ["Brentford", "Brighton", "Bournemouth"]
mask = df["HomeTeam"].isin(clubs) | df["AwayTeam"].isin(clubs)
df_3b = df[mask]

print(df_3b.shape)
print(df_3b.head())
```

```python
cols_needed = ["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR",
               "HS", "AS", "HST", "AST", "HC", "AC", "HF", "AF", "HY", "AY", "HR", "AR"]

df_3b_clean = df_3b[cols_needed]
print(df_3b_clean.shape)
print(df_3b_clean.head())
```

```python
home = df_3b_clean.rename(columns={
    "HomeTeam": "Team", "AwayTeam": "Opponent",
    "FTHG": "GoalsFor", "FTAG": "GoalsAgainst"
})
home["Venue"] = "Home"

away = df_3b_clean.rename(columns={
    "AwayTeam": "Team", "HomeTeam": "Opponent",
    "FTAG": "GoalsFor", "FTHG": "GoalsAgainst"
})
away["Venue"] = "Away"

team_matches = pd.concat([home, away], ignore_index=True)
team_matches = team_matches[team_matches["Team"].isin(clubs)]

print(team_matches.shape)
print(team_matches[["Date","Team","Opponent","Venue","GoalsFor","GoalsAgainst","FTR"]].head(10))
```

```python
def get_points(row):
    if row["Venue"] == "Home":
        if row["FTR"] == "H": return 3
        elif row["FTR"] == "D": return 1
        else: return 0
    else:
        if row["FTR"] == "A": return 3
        elif row["FTR"] == "D": return 1
        else: return 0

team_matches["Points"] = team_matches.apply(get_points, axis=1)

print(team_matches[["Team","Opponent","Venue","FTR","Points"]].head(10))
print(team_matches.groupby("Team")["Points"].sum())
```

```python
def load_season(url, season_label):
    df = pd.read_csv(url)
    mask = df["HomeTeam"].isin(clubs) | df["AwayTeam"].isin(clubs)
    df_3b = df[mask][cols_needed]

    home = df_3b.rename(columns={"HomeTeam":"Team","AwayTeam":"Opponent","FTHG":"GoalsFor","FTAG":"GoalsAgainst"})
    home["Venue"] = "Home"
    away = df_3b.rename(columns={"AwayTeam":"Team","HomeTeam":"Opponent","FTAG":"GoalsFor","FTHG":"GoalsAgainst"})
    away["Venue"] = "Away"

    tm = pd.concat([home, away], ignore_index=True)
    tm = tm[tm["Team"].isin(clubs)]
    tm["Points"] = tm.apply(get_points, axis=1)
    tm["Season"] = season_label
    return tm

season_2223 = load_season("https://football-data.co.uk/mmz4281/2223/E0.csv", "2022-23")
season_2324 = load_season("https://football-data.co.uk/mmz4281/2324/E0.csv", "2023-24")
season_2425 = load_season("https://football-data.co.uk/mmz4281/2425/E0.csv", "2024-25")

all_seasons = pd.concat([season_2223, season_2324, season_2425], ignore_index=True)

print(all_seasons.shape)
print(all_seasons.groupby(["Season","Team"])["Points"].sum())
```

```python
contenders = {
    "2022-23": ["Man City", "Arsenal", "Man United", "Newcastle"],
    "2023-24": ["Man City", "Arsenal", "Liverpool", "Aston Villa"],
    "2024-25": ["Liverpool", "Arsenal", "Man City", "Chelsea"]
}

def is_vs_contender(row):
    return row["Opponent"] in contenders[row["Season"]]

all_seasons["VsContender"] = all_seasons.apply(is_vs_contender, axis=1)

summary = all_seasons.groupby(["Season","Team","VsContender"])["Points"].agg(["sum","count"])
print(summary)
```

```python
contenders = {
    "2022-23": ["Man City", "Arsenal", "Man United", "Newcastle"],
    "2023-24": ["Man City", "Arsenal", "Liverpool", "Aston Villa"],
    "2024-25": ["Liverpool", "Arsenal", "Man City", "Chelsea"]
}

def is_vs_contender(row):
    return row["Opponent"] in contenders[row["Season"]]

all_seasons["VsContender"] = all_seasons.apply(is_vs_contender, axis=1)

summary = all_seasons.groupby(["Season","Team","VsContender"])["Points"].agg(["sum","count"])
print(summary)
```

```python
summary = summary.reset_index()
summary["PPG"] = summary["sum"] / summary["count"]
print(summary)
```

```python
tight_season = all_seasons[
    (all_seasons["Season"] == "2023-24") &
    (all_seasons["Opponent"].isin(["Man City", "Arsenal"]))
]

print(tight_season[["Team","Opponent","Venue","GoalsFor","GoalsAgainst","FTR","Points"]])
```

```python
champ_runnerup = {
    "2022-23": ["Man City", "Arsenal"],
    "2023-24": ["Man City", "Arsenal"],
    "2024-25": ["Liverpool", "Arsenal"]
}

def vs_top2(row):
    return row["Opponent"] in champ_runnerup[row["Season"]]

top2_matches = all_seasons[all_seasons.apply(vs_top2, axis=1)]

print(top2_matches[["Season","Team","Opponent","Venue","FTR","Points"]])
print(top2_matches.groupby("Team")["Points"].sum())
```
